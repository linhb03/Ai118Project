```python
!pip install language-tool-python

import language_tool_python

def analyze_and_correct(answer):
    tool = language_tool_python.LanguageTool('en-US')
    matches = tool.check(answer)
    mistakes = []
    for match in matches:
        mistakes.append({
            'mistake': answer[match.offset:match.offset + match.errorLength],
            'message': match.message,
            'suggestion': match.replacements
        })
    corrected = language_tool_python.utils.correct(answer, matches)
    explanations = []
    for m in mistakes:
        explanation = f"Original: '{m['mistake']}'. {m['message']} Suggested: {m['suggestion']}"
        explanations.append(explanation)
    tip = "Always check that your sentences have a clear subject, verb, and object. Read your writing aloud to catch missing words or awkward grammar."
    return mistakes, corrected, explanations, tip

# Student answer
student_answer = "English is difficult language because have many grammar. I try learn but is hard."

mistakes, corrected, explanations, tip = analyze_and_correct(student_answer)

print("Mistakes and Explanations:")
for expl in explanations:
    print(expl)

print("\nCorrected Answer:")
print(corrected)

print("\nTip to Avoid Mistakes:")
print(tip)
```

```python
# Step 1: Generate top 5 typical issues
def typical_issues(personalities):
    issues = {
        "Busy Professional": "Account access problems",
        "Tech-Savvy Millennial": "App crashes or bugs",
        "Senior Citizen": "Difficulty understanding features",
        "Frequent Traveler": "International usage/roaming issues",
        "Small Business Owner": "Billing/invoice discrepancies"
    }
    return [issues[p] for p in personalities]

# Step 2: Scripted empathetic statements and clarifying questions
def scripts_and_questions(issues):
    script_q = []
    for issue in issues:
        if issue == "Account access problems":
            script_q.append({
                "opening": "I'm sorry you're having trouble accessing your account. I understand how urgent this is for you.",
                "question": "Can you tell me if you’re seeing any error messages when you try to log in?"
            })
        elif issue == "App crashes or bugs":
            script_q.append({
                "opening": "I apologize for the inconvenience caused by the app crashing. Your experience matters to us.",
                "question": "Could you share which device and operating system you’re using?"
            })
        elif issue == "Difficulty understanding features":
            script_q.append({
                "opening": "I completely understand how navigating new features can be confusing. I'm here to help.",
                "question": "Is there a specific feature or button you’re struggling with?"
            })
        elif issue == "International usage/roaming issues":
            script_q.append({
                "opening": "Traveling should be easy, and I’m sorry to hear about your roaming issue.",
                "question": "Are you currently abroad, or did this happen during a previous trip?"
            })
        elif issue == "Billing/invoice discrepancies":
            script_q.append({
                "opening": "I understand how important accurate billing is for you. Let’s get this sorted out.",
                "question": "Could you let me know which invoice or transaction seems incorrect?"
            })
    return script_q

# Step 3: Final resolution script
def resolution_script():
    return (
        "Thank you for sharing those details. I’ve reviewed your issue and taken steps to resolve it. "
        "You should now have access to your account/app, or the billing discrepancy has been corrected. "
        "If you have any further questions or if the issue persists, please let us know—we’re always here to help!"
    )

# Simulation
personalities = ["Busy Professional", "Tech-Savvy Millennial", "Senior Citizen", "Frequent Traveler", "Small Business Owner"]
issues = typical_issues(personalities)
scripts = scripts_and_questions(issues)
final_resolution = resolution_script()

print("Top 5 Issues:")
for i, issue in enumerate(issues, 1):
    print(f"{i}. {issue}")

print("\nEmpathetic Openings & Clarifying Questions:")
for idx, s in enumerate(scripts, 1):
    print(f"{idx}. Opening: {s['opening']}")
    print(f"   Clarifying Question: {s['question']}")

print("\nFinal Resolution Script:")
print(final_resolution)
```
